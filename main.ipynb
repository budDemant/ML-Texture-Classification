### loadData.ipynb Defines the Following:
##### Library Imports
- `from pathlib import Path`: *Used to resolve filesystem paths.*
- `from collections import defaultdict`: *Used for variable creation.*
- `import matplotlib.pyplot as plt`: *Used for displaying images if VERBOSE_LOGS set to true.*
- `import numpy as np`: *Used for load_image(...) function.*
- `import pandas as pd`: *Used for variable creation.*
- `from PIL import Image`: *Used for load_image(...) function.*
- `from sklearn.preprocessing import LabelEncoder`: *Used for labeling in material_datasets variable.*
- `from skimage.color import rgb2gray`: *Used for load_image(...) function.*
##### Variables:
- **image_files**: *A list of all images' paths within the dataset.*
- **images_by_material**: *A dictionary of images sorted by material.*
- **dataset_records**: *A list of all records in the dataset, each entry has the following properties: "image_path", "material", "sample", "filename"*
- **df**: *A Pandas dataframe of dataset_records.*
- **material_datasets**: *A dictionary of all materials, each material has the following properties: "paths" (list of all image paths), "samples" (list of all samples), "labels" (list of material labels).*

##### Functions:
- **load_image(image_path, grayscale=False)**: *Loads a specified image, optionally grayscaling; returns a numpy.ndarray (Image array).*
